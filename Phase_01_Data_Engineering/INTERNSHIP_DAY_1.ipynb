### Task 3: Department Wise Analysis

```python
print("Count of students in each department:")
print(df['department'].value_counts())
```

### Task 4: Score Analysis

```python
print(f"Highest Math Score: {df['math_score'].max()}")
print(f"Lowest Math Score: {df['math_score'].min()}")
print(f"Average Math Score: {df['math_score'].mean():.2f}")
print(f"Average Programming Score: {df['programming_score'].mean():.2f}")
print(f"Average Attendance Percentage: {df['attendance_percentage'].mean():.2f}")
```

## Answers to Practice Questions

### Question 1: How many students are from the Electronics department?

```python
electronics_students_count = df[df['department'] == 'Electronics'].shape[0]
print(f"Number of students from the Electronics department: {electronics_students_count}")
```

### Question 2: What is the average attendance of female students?

```python
avg_attendance_female = df[df['gender'] == 'Female']['attendance_percentage'].mean()
print(f"Average attendance of female students: {avg_attendance_female:.2f}%")
```

### Question 3: Who is the student with the lowest programming score?

```python
lowest_programming_student = df.loc[df['programming_score'].idxmin()]
print(f"Student with the lowest programming score: {lowest_programming_student['name']} (Score: {lowest_programming_student['programming_score']})")
```

### Question 4: How many students scored above 80 in math?

```python
students_above_80_math = df[df['math_score'] > 80].shape[0]
print(f"Number of students who scored above 80 in math: {students_above_80_math}")
```

### Question 5: What is the average total_score for students with attendance above 90%?

```python
avg_total_score_high_attendance = df[df['attendance_percentage'] > 90]['total_score'].mean()
print(f"Average total score for students with attendance above 90%: {avg_total_score_high_attendance:.2f}")
```
